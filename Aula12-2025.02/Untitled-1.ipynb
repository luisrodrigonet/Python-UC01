### Declarando a classe Carro

```python
class Carro:

    def __init__(self,marca, modelo, ano ):
        self.marca = marca
        self.modelo = modelo
        self.ano = ano
        self.cor = ""
        self._garantia = True
        self._ligado = False

    def ligar(self):
        if self._ligado :
            print (f"\n\t [ERRO] Já está ligado \n")
        else :
            self._ligado=True
            print (f"\t\t [OK] O carro está ligado !!!")
            
    def desligar(self):
        if self._ligado :
            self._ligado=False
            print (f"\t\t [OK] O carro foi DESligado !!!")
        else :
            print (f"\t\t [ERRO] Já está DESligado \n")

    def exibir_info(self):
        if self._ligado :
            status = "Ligado"
        else: 
            status = "Desligado"        
        print(f"{self.marca} - {self.modelo} está {status}")
```

```python
meu_carro = Carro("WV", "Gol", 2013)
```

```python
print(meu_carro)
```

```python
meu_carro.exibir_info ()
```

```python
meu_carro.ligar()
```

```python
meu_carro.exibir_info()
```

```python
meu_carro.desligar()
```

```python
meu_carro.exibir_info ()
```
