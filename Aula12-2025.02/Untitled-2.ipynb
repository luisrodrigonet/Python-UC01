### Declarando a classe Carro

```python
class Carro:

    def __init__(self,marca, modelo, ano ):
        self.marca = marca
        self.modelo = modelo
        self.ano = ano
        self.cor = ""
        self._garantia = True
        self._ligado = False

    def ligar(self):
        if self._ligado :
            print (f"\n\t [ERRO] Já está ligado \n")
        else :
            self._ligado=True
            print (f"\t\t [OK] O carro está ligado !!!")
            
    def desligar(self):
        if self._ligado :
            self._ligado=False
            print (f"\t\t [OK] O carro foi DESligado !!!")
        else :
            print (f"\t\t [ERRO] Já está DESligado \n")

    def exibir_info(self):
        if self._ligado :
            status = "Ligado"
        else: 
            status = "Desligado"        
        print(f"{self.marca} - {self.modelo} está {status}")
```

### Criando os carros

```python
mc1 = Carro("WV", "Gol", 2013)
mc2 = Carro("Fiat", "Uno", 2020)
mc3 = Carro("WV", "Golf", 2025)
```

### Criando dicionário

```python
colecao={}

colecao["kpl2132"]=mc1
colecao["xyz"]=mc2
colecao["abc987"]=mc3
```

### Acessando os carros que estão no dicionário

```python
colecao["kpl2132"].exibir_info()
colecao["kpl2132"].ligar()
colecao["kpl2132"].exibir_info()
```
